```python
import pandas as pd

# Create Employee Dataset
data = {
    "Employee_ID": ["101", "102", "103", "104", "105"],
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi"],
    "Age": ["25", "30", "28", "35", "40"],
    "Salary": ["25000.50", "30000.00", "28000.75", "35000.00", "40000.50"],
    "Joining_Date": [
        "2023-01-10",
        "2022-05-15",
        "2021-08-20",
        "2020-03-12",
        "2019-11-25"
    ]
}

df = pd.DataFrame(data)

# Display Original Dataset
print("1. ORIGINAL DATASET")
print(df)

# Check Original Data Types
print("\n2. ORIGINAL DATA TYPES")
print(df.dtypes)

# Convert Data Types
df["Employee_ID"] = df["Employee_ID"].astype("int64")
df["Age"] = df["Age"].astype("int64")
df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")
df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"], errors="coerce"
)

# Display Corrected Data Types
print("\n3. CORRECTED DATA TYPES")
print(df.dtypes)

# Display Cleaned Dataset
print("\n4. CLEANED DATASET")
print(df)

# Dataset Information
print("\n5. DATASET INFORMATION")
df.info()

# Statistical Summary
print("\n6. STATISTICAL SUMMARY")
print(df.describe())

# Check Missing Values
print("\n7. MISSING VALUES")
print(df.isnull().sum())

# Observations
print("\n8. OBSERVATIONS")
print("1. Employee_ID converted from object to integer.")
print("2. Age converted from object to integer.")
print("3. Salary converted from object to float.")
print("4. Joining_Date converted from object to datetime.")
print("5. Name remains object because it contains text.")
print("6. All columns have been checked for missing values.")
print("7. Correct data types improve data analysis.")
```
